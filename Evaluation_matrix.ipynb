# MoodSyncAI Model Performance Analysis

This notebook demonstrates how to visualize performance-related metrics for MoodSyncAI. It includes example charts for emotion confidence, sentiment confidence, mismatch analysis, and a small demo of how the system outputs can be tracked.

## 1. Import Required Libraries

Import the libraries needed for data handling and visualization.

```python
import pandas as pd
import altair as alt
import numpy as np

print('Libraries loaded successfully')
```

## 2. Example Model Output Data

Because MoodSyncAI performs live emotion and sentiment detection, this notebook uses example data to show how performance metrics and charts can be generated. Replace this with real prediction results when available.

```python
example_data = pd.DataFrame([
    {'sample': 'session_01', 'emotion': 'Happy', 'emotion_confidence': 91.2, 'sentiment': 'POSITIVE', 'sentiment_confidence': 0.88, 'mismatch': False},
    {'sample': 'session_02', 'emotion': 'Sad', 'emotion_confidence': 79.4, 'sentiment': 'POSITIVE', 'sentiment_confidence': 0.72, 'mismatch': True},
    {'sample': 'session_03', 'emotion': 'Neutral', 'emotion_confidence': 84.0, 'sentiment': 'NEGATIVE', 'sentiment_confidence': 0.65, 'mismatch': False},
    {'sample': 'session_04', 'emotion': 'Angry', 'emotion_confidence': 88.3, 'sentiment': 'POSITIVE', 'sentiment_confidence': 0.93, 'mismatch': True},
    {'sample': 'session_05', 'emotion': 'Happy', 'emotion_confidence': 96.5, 'sentiment': 'POSITIVE', 'sentiment_confidence': 0.96, 'mismatch': False},
])
example_data
```

## 3. Simple Performance Metrics

Compute summary statistics from the example outputs. These metrics can be extended for formal model evaluation with labeled test data.

```python
summary_metrics = {
    'average_emotion_confidence': example_data['emotion_confidence'].mean(),
    'average_sentiment_confidence': example_data['sentiment_confidence'].mean(),
    'mismatch_rate_percent': example_data['mismatch'].mean() * 100,
    'positive_sentiment_rate_percent': (example_data['sentiment'] == 'POSITIVE').mean() * 100,
}
pd.DataFrame([summary_metrics])
```

## 4. Evaluation Metrics: Precision, Recall, F1

For a classification problem, precision, recall, and F1 score help measure how well the model predictions match true labels. Here we compute these metrics for text sentiment using example ground truth values.

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score, classification_report

# Example ground truth sentiment labels for evaluation
example_data['true_sentiment'] = ['POSITIVE', 'NEGATIVE', 'NEGATIVE', 'POSITIVE', 'POSITIVE']

# Convert labels to binary values for metric calculation
label_map = {'NEGATIVE': 0, 'POSITIVE': 1}

y_true = example_data['true_sentiment'].map(label_map)
y_pred = example_data['sentiment'].map(label_map)

metrics = {
    'accuracy': accuracy_score(y_true, y_pred),
    'precision': precision_score(y_true, y_pred),
    'recall': recall_score(y_true, y_pred),
    'f1_score': f1_score(y_true, y_pred),
}

print('Classification Report:')
print(classification_report(y_true, y_pred, target_names=['NEGATIVE', 'POSITIVE']))
pd.DataFrame([metrics])
```

## 4. Emotion Confidence Chart

Visualize emotion model confidence across sessions.

```python
emotion_chart = alt.Chart(example_data).mark_bar(color='#57de63').encode(
    x=alt.X('sample:N', title='Session'),
    y=alt.Y('emotion_confidence:Q', title='Emotion Confidence (%)'),
    tooltip=['sample', 'emotion', 'emotion_confidence']
).properties(height=300, width=650)
emotion_chart
```

## 5. Sentiment Confidence Chart

Show the sentiment model confidence for the same sessions.

```python
sentiment_chart = alt.Chart(example_data).mark_line(point=True, color='#fb7185').encode(
    x=alt.X('sample:N', title='Session'),
    y=alt.Y('sentiment_confidence:Q', title='Sentiment Confidence'),
    tooltip=['sample', 'sentiment', 'sentiment_confidence']
).properties(height=300, width=650)
sentiment_chart
```

## 6. Mismatch Summary

A mismatch means the system detected positive verbal sentiment while the facial emotion is in a negative category.

```python
mismatch_summary = example_data.groupby('mismatch').size().reset_index(name='count')
mismatch_summary['status'] = mismatch_summary['mismatch'].map({True: 'Mismatch', False: 'Match'})
mismatch_chart = alt.Chart(mismatch_summary).mark_arc(innerRadius=50).encode(
    theta=alt.Theta('count:Q'),
    color=alt.Color('status:N', title='Match Status'),
    tooltip=['status', 'count']
).properties(height=300, width=400)
mismatch_chart
```
